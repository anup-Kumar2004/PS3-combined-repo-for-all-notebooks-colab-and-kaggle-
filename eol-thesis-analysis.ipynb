```python
import os
import pickle
import numpy as np
import pandas as pd

BASE = "/kaggle/input/datasets/nikhilrajiv12/battery-degradation-dataset-xai"
HUST_DIR = os.path.join(BASE, "HUST")
MATR_DIR = os.path.join(BASE, "MATR")

NOMINAL_CAPACITY = 1.1
EOL_THRESHOLD = 0.88  # 80% of 1.1 Ah

def process_battery(pkl_path, dataset_source):
    with open(pkl_path, "rb") as f:
        data = pickle.load(f)

    cell_id = data["cell_id"]
    cycles = data["cycle_data"]

    records = []
    for c in cycles:
        cycle_num = c["cycle_number"]

        discharge_list = c["discharge_capacity_in_Ah"]
        charge_list = c["charge_capacity_in_Ah"]

        if not discharge_list or not charge_list:
            continue

        discharge_cap = discharge_list[-1]
        charge_cap = charge_list[-1]

        if discharge_cap <= 0 or charge_cap <= 0:
            continue

        records.append({
            "cell_id": cell_id,
            "dataset": dataset_source,
            "cycle_number": cycle_num,
            "discharge_capacity": discharge_cap,
            "charge_capacity": charge_cap,
        })

    if not records:
        return None

    df = pd.DataFrame(records)
    df = df.sort_values("cycle_number").reset_index(drop=True)

    # SOH from first cycle discharge (more reliable than nominal)
    initial_capacity = df["discharge_capacity"].iloc[0]
    df["soh"] = df["discharge_capacity"] / initial_capacity

    # find EOL cycle (first cycle where SOH drops below 80%)
    eol_mask = df["soh"] < 0.80
    if eol_mask.any():
        eol_cycle = df.loc[eol_mask.idxmax(), "cycle_number"]
        df = df[df["cycle_number"] <= eol_cycle].copy()
        reached_eol = True
    else:
        eol_cycle = df["cycle_number"].max()
        reached_eol = False

    total_life = len(df)

    # normalized RUL: 1.0 at start, 0.0 at EOL
    df["rul_normalized"] = 1.0 - (df.index / (total_life - 1)) if total_life > 1 else 0.0

    df["reached_eol"] = reached_eol
    df["total_life_cycles"] = total_life
    df["coulombic_efficiency"] = df["discharge_capacity"] / df["charge_capacity"]

    return df

print("Processing HUST...")
hust_frames = []
hust_files = sorted([f for f in os.listdir(HUST_DIR) if f.endswith(".pkl")])
for fname in hust_files:
    df = process_battery(os.path.join(HUST_DIR, fname), "HUST")
    if df is not None:
        hust_frames.append(df)

print(f"  Processed {len(hust_frames)} HUST batteries")

print("Processing MATR...")
matr_frames = []
matr_files = sorted([f for f in os.listdir(MATR_DIR) if f.endswith(".pkl")])
for fname in matr_files:
    df = process_battery(os.path.join(MATR_DIR, fname), "MATR")
    if df is not None:
        matr_frames.append(df)

print(f"  Processed {len(matr_frames)} MATR batteries")

hust_df = pd.concat(hust_frames, ignore_index=True)
matr_df = pd.concat(matr_frames, ignore_index=True)
combined_df = pd.concat([hust_df, matr_df], ignore_index=True)

print()
print(f"HUST total rows:    {len(hust_df)}")
print(f"MATR total rows:    {len(matr_df)}")
print(f"Combined total rows:{len(combined_df)}")
print()
print("HUST EOL reached:", hust_df.groupby("cell_id")["reached_eol"].first().value_counts().to_dict())
print("MATR EOL reached:", matr_df.groupby("cell_id")["reached_eol"].first().value_counts().to_dict())
```

```python
# fix SOH outliers — cap SOH at 1.10 maximum (10% above nominal is realistic)
# anything above that is a measurement artifact in early cycles
final_df = final_df[final_df["soh"] <= 1.10].copy()

print("After SOH outlier removal:")
print(f"  Total rows:     {len(final_df)}")
print(f"  Total batteries:{final_df['cell_id'].nunique()}")
print()
print("SOH distribution after cleaning:")
print(final_df.groupby("dataset")["soh"].describe().round(4))
print()

# add rolling features per battery
print("Adding rolling features...")

def add_rolling_features(df):
    df = df.sort_values("cycle_number").copy()

    df["capacity_fade"] = df["soh"].diff().fillna(0)
    df["soh_roll5"] = df["soh"].rolling(window=5, min_periods=1).mean()

    # slope of SOH over last 10 cycles
    def slope(x):
        if len(x) < 2:
            return 0.0
        y = x.values
        x_idx = np.arange(len(y))
        return np.polyfit(x_idx, y, 1)[0]

    df["soh_trend10"] = df["soh"].rolling(window=10, min_periods=2).apply(slope, raw=False)
    df["soh_trend10"] = df["soh_trend10"].fillna(0)

    return df

final_df = final_df.groupby("cell_id", group_keys=False).apply(add_rolling_features)
final_df = final_df.reset_index(drop=True)

print("Rolling features added.")
print()
print("Sample of final features (first 3 rows of HUST_1-1):")
cols = ["cell_id", "cycle_number", "soh", "coulombic_efficiency",
        "capacity_fade", "soh_roll5", "soh_trend10", "rul_normalized"]
print(final_df[final_df["cell_id"] == "HUST_1-1"][cols].head(3).to_string(index=False))
print()
print("Sample of final features (last 3 rows of HUST_1-1):")
print(final_df[final_df["cell_id"] == "HUST_1-1"][cols].tail(3).to_string(index=False))
print()

# verify no nulls
print("Null check:")
print(final_df[cols].isnull().sum())
```

```python
# the issue is only in the very last cycle of a few batteries
# where there is a sudden abnormal drop
# fix: remove the last cycle if it drops more than 5% from the previous cycle

def remove_anomalous_last_cycle(df):
    if len(df) < 3:
        return df
    
    df = df.sort_values("cycle_number").copy()
    last_soh = df["soh"].iloc[-1]
    prev_soh = df["soh"].iloc[-2]
    
    # if last cycle drops more than 5% from previous, remove it
    if (prev_soh - last_soh) > 0.05:
        df = df.iloc[:-1].copy()
        # recalculate rul_normalized after removing last cycle
        total = len(df)
        df["rul_normalized"] = 1.0 - (
            pd.Series(range(total)) / (total - 1)
        ).values if total > 1 else 0.0
    
    return df

print("Removing anomalous last cycles...")
final_df = final_df.groupby(
    "cell_id", group_keys=False
).apply(remove_anomalous_last_cycle, include_groups=True)
final_df = final_df.reset_index(drop=True)

print("Done.")
print()

# verify the fix
for bat in ["MATR_b4c35", "MATR_b4c38", "MATR_b4c25"]:
    bat_df = final_df[final_df["cell_id"] == bat]
    if len(bat_df) > 0:
        print(f"{bat}: {len(bat_df)} cycles, "
              f"min SOH={bat_df['soh'].min():.4f}, "
              f"last SOH={bat_df['soh'].iloc[-1]:.4f}, "
              f"last RUL={bat_df['rul_normalized'].iloc[-1]:.4f}")
print()

# final quality summary
print("=== FINAL DATASET QUALITY SUMMARY ===")
print()
print(f"Total batteries: {final_df['cell_id'].nunique()}")
print(f"Total rows:      {len(final_df)}")
print()

per_battery = final_df.groupby(["cell_id", "dataset"]).agg(
    total_cycles=("cycle_number", "count"),
    min_soh=("soh", "min"),
    max_soh=("soh", "max"),
    final_rul=("rul_normalized", "last"),
).reset_index()

print("Per-battery stats by dataset:")
print(per_battery.groupby("dataset").agg(
    batteries=("cell_id", "count"),
    min_cycles=("total_cycles", "min"),
    mean_cycles=("total_cycles", "mean"),
    max_cycles=("total_cycles", "max"),
    min_soh_overall=("min_soh", "min"),
    max_soh_overall=("max_soh", "max"),
).round(1).to_string())
print()

# confirm all batteries end at RUL=0
bad_rul = per_battery[per_battery["final_rul"] > 0.01]
if len(bad_rul) == 0:
    print("All batteries correctly end at RUL = 0.0")
else:
    print(f"WARNING: {len(bad_rul)} batteries do not end at RUL=0:")
    print(bad_rul[["cell_id", "final_rul"]])

print()

# save final clean version
output_path = "/kaggle/working/combined_eol_dataset_clean.csv"
final_df.to_csv(output_path, index=False)
print(f"Clean dataset saved to: {output_path}")
```

```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.metrics import r2_score, mean_absolute_error
import random
import warnings
warnings.filterwarnings("ignore")

df = pd.read_csv("/kaggle/working/combined_eol_dataset_clean.csv")

FEATURES = [
    "cycle_number",
    "soh",
    "coulombic_efficiency",
    "capacity_fade",
    "soh_roll5",
    "soh_trend10",
]
TARGET = "rul_normalized"

all_batteries = df["cell_id"].unique().tolist()
random.seed(42)
np.random.seed(42)

# fixed test set — 20 batteries held out permanently
# stratified: 10 HUST + 10 MATR
hust_batteries = df[df["dataset"] == "HUST"]["cell_id"].unique().tolist()
matr_batteries = df[df["dataset"] == "MATR"]["cell_id"].unique().tolist()

random.shuffle(hust_batteries)
random.shuffle(matr_batteries)

test_batteries = hust_batteries[:10] + matr_batteries[:10]
train_pool = [b for b in all_batteries if b not in test_batteries]

print(f"Fixed test set:  {len(test_batteries)} batteries "
      f"({sum(1 for b in test_batteries if 'HUST' in b)} HUST + "
      f"{sum(1 for b in test_batteries if 'MATR' in b)} MATR)")
print(f"Training pool:   {len(train_pool)} batteries")
print()

test_df = df[df["cell_id"].isin(test_batteries)]
X_test = test_df[FEATURES].values
y_test = test_df[TARGET].values

# training sizes to sweep
TRAINING_SIZES = [5, 10, 15, 20, 30, 40, 50, 60, 80, 100, 120, 145]
N_REPEATS = 20  # random subsets per size

print(f"Training sizes to sweep: {TRAINING_SIZES}")
print(f"Repeats per size: {N_REPEATS}")
print(f"Total model runs: {len(TRAINING_SIZES) * N_REPEATS}")
print()
print("Running experiment...")
print()

results = []

for n in TRAINING_SIZES:
    r2_scores = []
    mae_scores = []

    for repeat in range(N_REPEATS):
        subset = random.sample(train_pool, n)
        train_subset = df[df["cell_id"].isin(subset)]

        X_train = train_subset[FEATURES].values
        y_train = train_subset[TARGET].values

        model = XGBRegressor(
            n_estimators=200,
            max_depth=6,
            learning_rate=0.05,
            subsample=0.8,
            colsample_bytree=0.8,
            random_state=42,
            verbosity=0,
        )
        model.fit(X_train, y_train)

        y_pred = model.predict(X_test)
        r2_scores.append(r2_score(y_test, y_pred))
        mae_scores.append(mean_absolute_error(y_test, y_pred))

    results.append({
        "n_batteries": n,
        "r2_mean": np.mean(r2_scores),
        "r2_std": np.std(r2_scores),
        "r2_min": np.min(r2_scores),
        "r2_max": np.max(r2_scores),
        "mae_mean": np.mean(mae_scores),
        "mae_std": np.std(mae_scores),
    })

    print(f"  n={n:3d} batteries → "
          f"R²={np.mean(r2_scores):.4f} ± {np.std(r2_scores):.4f} | "
          f"MAE={np.mean(mae_scores):.4f} ± {np.std(mae_scores):.4f}")

results_df = pd.DataFrame(results)
results_df.to_csv("/kaggle/working/eol_sweep_results.csv", index=False)
print()
print("Results saved to /kaggle/working/eol_sweep_results.csv")
```

```python
from sklearn.metrics import r2_score, mean_absolute_error
from xgboost import XGBRegressor
import numpy as np
import random
import warnings
warnings.filterwarnings("ignore")

FEATURES_FULL = [
    "cycle_number", "soh", "coulombic_efficiency",
    "capacity_fade", "soh_roll5", "soh_trend10"
]
FEATURES_BASE = [
    "cycle_number", "soh", "coulombic_efficiency", "capacity_fade"
]
TARGET = "rul_normalized"

X_test_full = test_df[FEATURES_FULL].values
X_test_base = test_df[FEATURES_BASE].values
y_test = test_df[TARGET].values

SWEEP_SIZES = [5, 10, 20, 40, 80, 145]
N_REPEATS = 20

print("=== Ablation: Full features vs Base features (no rolling SOH) ===")
print()
print(f"{'n':>4} | {'Full R²':>10} {'±':>3} {'std':>6} | "
      f"{'Base R²':>10} {'±':>3} {'std':>6} | {'Δ R²':>8}")
print("-" * 65)

ablation_results = []

for n in SWEEP_SIZES:
    r2_full_list, r2_base_list = [], []

    for _ in range(N_REPEATS):
        subset = random.sample(train_pool, n)
        train_sub = df[df["cell_id"].isin(subset)]

        model = XGBRegressor(n_estimators=200, max_depth=6,
                             learning_rate=0.05, subsample=0.8,
                             colsample_bytree=0.8, random_state=42,
                             verbosity=0)

        model.fit(train_sub[FEATURES_FULL].values,
                  train_sub[TARGET].values)
        r2_full_list.append(r2_score(y_test,
                            model.predict(X_test_full)))

        model.fit(train_sub[FEATURES_BASE].values,
                  train_sub[TARGET].values)
        r2_base_list.append(r2_score(y_test,
                            model.predict(X_test_base)))

    r2_full = np.mean(r2_full_list)
    r2_base = np.mean(r2_base_list)

    ablation_results.append({
        "n_batteries": n,
        "r2_full_mean": r2_full,
        "r2_full_std": np.std(r2_full_list),
        "r2_base_mean": r2_base,
        "r2_base_std": np.std(r2_base_list),
        "delta_r2": r2_full - r2_base,
    })

    print(f"{n:>4} | {r2_full:>10.4f} {'±':>3} "
          f"{np.std(r2_full_list):>6.4f} | "
          f"{r2_base:>10.4f} {'±':>3} "
          f"{np.std(r2_base_list):>6.4f} | "
          f"{r2_full - r2_base:>+8.4f}")

print()
print("=== Cross-dataset validation ===")
print()

hust_test = test_df[test_df["dataset"] == "HUST"]
matr_test = test_df[test_df["dataset"] == "MATR"]

for n in [40, 145]:
    r2_hust_list, r2_matr_list = [], []

    for _ in range(N_REPEATS):
        subset = random.sample(train_pool, n)
        train_sub = df[df["cell_id"].isin(subset)]

        model = XGBRegressor(n_estimators=200, max_depth=6,
                             learning_rate=0.05, subsample=0.8,
                             colsample_bytree=0.8, random_state=42,
                             verbosity=0)
        model.fit(train_sub[FEATURES_FULL].values,
                  train_sub[TARGET].values)

        r2_hust_list.append(r2_score(
            hust_test[TARGET].values,
            model.predict(hust_test[FEATURES_FULL].values)
        ))
        r2_matr_list.append(r2_score(
            matr_test[TARGET].values,
            model.predict(matr_test[FEATURES_FULL].values)
        ))

    print(f"n={n} batteries:")
    print(f"  HUST test R²: {np.mean(r2_hust_list):.4f} "
          f"± {np.std(r2_hust_list):.4f}")
    print(f"  MATR test R²: {np.mean(r2_matr_list):.4f} "
          f"± {np.std(r2_matr_list):.4f}")
    print()

ablation_df = pd.DataFrame(ablation_results)
ablation_df.to_csv("/kaggle/working/ablation_results.csv", index=False)
print("Ablation results saved.")
```
